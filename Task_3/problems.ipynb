```python
import numpy as np
```

```python
# P1
arr = np.eye(3) * 7

print(arr)
```

```python
# P2
arr = np.zeros((5, 5))

arr[0, :] = 1
arr[-1, :] = 1

arr[2, 2] = 5

print(arr)
```

```python
# P3
m = int(input("Enter m: "))
n = int(input("Enter n: "))

row = np.arange(1, n + 1)

arr = np.broadcast_to(row, (m, n))

print(arr)
```

```python
# P4
arr = np.arange(1, 21)

result = arr[(arr % 2 == 0) & (arr > 10)]

print(result)
```

```python
# P5
arr = np.array([12, 7, 25, 18, 30, 10, 22, 15])

print("Sum:", np.sum(arr))
print("Average:", np.mean(arr))
print("Maximum:", np.max(arr))
print("Minimum:", np.min(arr))
print("Index of maximum:", np.argmax(arr))
```
