# Getting and Knowing your Data

### Step 1. Import the necessary libraries

```python
import pandas as pd
```

### Step 2. Import the dataset & Assign it to a variable called chipo.

```python
chipo = pd.read_csv("chipotle.csv")
```

### Step 4. See the first 10 entries

```python
chipo.head(10)
```

### Step 5. What is the number of columns in the dataset?

```python
chipo.shape[1]
```

### Step 6. Print a summary of the DataFrame

```python
chipo.info()
```

### Step 7. Print the name of all the columns.

```python
chipo.columns
```

### Step 8.Display only the item_name column.

```python
chipo["item_name"]
```

### Step 9. Use `.loc` to display the first 5 rows for only two columns: item_name and item_price

```python
chipo.loc[0:4, ["item_name", "item_price"]]
```

### Step 10. Sort by the name of the item

```python
chipo.sort_values(by="item_name")
```

### Step 11. Which was the most-ordered item?

```python
chipo["item_name"].value_counts().head(1)
```

### Step 12. For the most-ordered item, how many items were ordered?

```python
most_ordered = chipo["item_name"].value_counts().idxmax()

chipo.loc[chipo["item_name"] == most_ordered, "quantity"].sum()
```

### Step 13. What was the most ordered item in the choice_description column?

```python
chipo["choice_description"].value_counts().head(1)
```

### Step 14. How many items were orderd in total?

```python
chipo["quantity"].sum()
```

### Step 15. Retrieve all orders where the item_name is exactly 'Chicken Bowl'

```python
chipo[chipo["item_name"] == "Chicken Bowl"]
```

### Step 16. Find all orders where the quantity is greater than 2.

```python
chipo[chipo["quantity"] > 2]
```

### Step 17. How many products cost more than $10.00?

```python
chipo["item_price"] = chipo["item_price"].str.replace("$", "", regex=False).astype(float)

(chipo["item_price"] > 10).sum()
```

### Step 18.Set the order_id column as the index of your DataFrame.

```python
chipo.set_index("order_id", inplace=True)
```

#### 18.a. Display the first 5 rows to see how the new index looks.

```python
chipo.head(5)
```

#### 18.b. Reset the index back to the default integer index.

```python
chipo.reset_index(inplace=True)
```

### Step 19. Rename the column item_name to product_name.

```python
chipo.rename(columns={"item_name": "product_name"}, inplace=True)
```

### Step 20. Convert all text in the choice_description column to lowercase letters.

```python
chipo["choice_description"] = chipo["choice_description"].str.lower()
```

### Step 21. Create a new column called is_expensive and set it to True if the price is above 12, and False otherwise.

```python
chipo["is_expensive"] = chipo["item_price"] > 12
```

### Step 22. Delete the choice_description column from the DataFrame

```python
chipo.drop("choice_description", axis=1, inplace=True)
```
